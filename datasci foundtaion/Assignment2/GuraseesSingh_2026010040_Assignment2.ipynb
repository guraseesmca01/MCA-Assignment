```python
#Ques17: point location using match case
x=int(input("Enter x coordinate:"))
y=int(input("Enter y coordinate:"))
match (x,y):
  case (_,_) if x==0 and y==0:
    print("Origin")
  case (_,_) if x>0 and y==0:
    print("Positive X-axis")
  case (_,_) if x<0 and y==0:
    print("Negative X-axis")
  case (_,_) if y>0 and x==0:
    print("Positive Y-axis")
  case (_,_) if y<0 and x==0:
    print("Negative Y-axis")
  case (_,_) if x>0 and y>0:
    print("First Quadrant")
  case (_,_) if x<0 and y>0:
    print("Second Quadrant")
  case (_,_) if x<0 and y<0:
    print("Third Quadrant")
  case (_,_0) if x>0 and y<0:
    print("Fourth Quadrant")

```

```python
#Ques20: number guessing game using while else
secret_number=4

attempts=0

while attempts<5:
  input_number=int(input("Enter a number:"))
  if(input_number==secret_number):
      print("Correct!")
      break
  elif(input_number>secret_number):
    print("Too High")
    attempts+=1
  elif(input_number<secret_number):
    print("Too Low")
    attempts+=1
else:
    print("All attempts are exhausted")
```

```python
#Ques47: taking 3*3 matrix from user and printing element at index
matrix=[]

for i in range(3):
  row=[]
  for j in range(3):
    num=int(input(f"Enter element at [{i}][{j}]:"))
    row.append(num)
  matrix.append(row)

print("\nThe matrix you entered:")

for i in range(3):
    for j in range(3):
        print(matrix[i][j], end=" ")
    print()

print("\nElement at row index 2, column index 1:", matrix[2][1])
```
